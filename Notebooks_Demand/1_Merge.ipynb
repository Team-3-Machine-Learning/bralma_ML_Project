# Mergin all files
The second dataset exist out of multiple files. We will merge them into one file so it is easier to use.

```python
import pandas as pd
import glob
import os

path = "../Dataset2_Demand/Raw/"
path_out = "../Dataset2_Demand/1_Elec_Demand_Merged.csv"

all_files = glob.glob(os.path.join(path, "*.csv"))
dfs = []

for file in sorted(all_files):
    df = pd.read_csv(file)
    
    year = os.path.basename(file).split('_')[-1].split('.')[0]
    df['Year'] = int(year)
    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)
combined.to_csv(path_out, index=False)
```
