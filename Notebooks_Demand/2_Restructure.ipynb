# Restructuring dataset 2
### Renaming the columns
The column names of dataset 2 contain capital letters. We are going to rename them: removing capital letters and replacing them with regular letters.

```python
import pandas as pd

path = "../Dataset2_Demand/1_Dataset2_Demand_Merged.csv"
path_out = "../Dataset2_Demand/2_Dataset2_Demand_Restructured.csv"

df = pd.read_csv(path)
df.columns
```

```python
df.columns = [col.lower() for col in df.columns]
df.columns
```

```python
df.to_csv(path_out, index=False)
```
