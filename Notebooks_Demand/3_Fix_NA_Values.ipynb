# Fixing NA values in the dataset
We need to fix all null values in our dataset.

```python
import pandas as pd

path = "../Dataset2_Demand/2_Elec_Demand_Restructured.csv"
path_out_mean = "../Dataset2_Demand/3_Elec_Demand_mean.csv"
path_out_zero = "../Dataset2_Demand/3_Elec_Demand_zero.csv"
path_out_na = "../Dataset2_Demand/3_Elec_Demand_na.csv"
```

### Using the mean
Our first tactic is to replace all NA values using the mean value of that column.

```python
df = pd.read_csv(path)
means = df.mean(numeric_only=True)
df_filled = df.fillna(means)
df_filled.to_csv(path_out_mean, index=False)

diff = df.ne(df_filled)
rows_changed = diff.any(axis=1).sum()
column_changes = diff.sum().sort_values(ascending=False)

top_3_columns = column_changes.head(3)

print("Rows changed:", rows_changed)
if rows_changed != 0:
    print("Top 3 changed columns:")
    print(top_3_columns)
```

### Using zero
Our second tactic is to replace all null values with zero.

```python
df = pd.read_csv(path)
df_filled = df.fillna(0)
df_filled.to_csv(path_out_zero, index=False)

diff = df.ne(df_filled)
rows_changed = diff.any(axis=1).sum()
column_changes = diff.sum().sort_values(ascending=False)

top_3_columns = column_changes.head(3)

print("Rows changed:", rows_changed)
if rows_changed != 0:
    print("Top 3 changed columns:")
    print(top_3_columns)
```

### Using NA
Our third tactic is to replace all null values with NA.

```python
df = pd.read_csv(path)
df = df.fillna('NA')
df.to_csv(path_out_na, index=False)

diff = df.ne(df_filled)
rows_changed = diff.any(axis=1).sum()
column_changes = diff.sum().sort_values(ascending=False)

top_3_columns = column_changes.head(3)

print("Rows changed:", rows_changed)
if rows_changed != 0:
    print("Top 3 changed columns:")
    print(top_3_columns)
```

We will continue using the NA method, because we can easily drop columns later on to make our subsets. The mean method does totally not work for our data, because we have some columns that are NA for almost the whole dataset, beside the last 5 years.
