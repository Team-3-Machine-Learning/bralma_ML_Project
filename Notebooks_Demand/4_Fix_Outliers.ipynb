# Fix outliers in dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

path = "../Dataset2_Demand/3_Elec_Demand_na.csv"
path_out = "../Dataset2_Demand/4_Elec_Demand_fix_outliers.csv"
```

```python
df = pd.read_csv(path)
# Code to detect and fix outliers
```

```python
# IQR method for full dataset

Q1 = df['england_wales_demand'].quantile(0.25)
Q3 = df['england_wales_demand'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['england_wales_demand'] < lower_bound) | (df['england_wales_demand'] > upper_bound)]
print(f"Outliers: {len(outliers)} rows")
```

```python
print(outliers[['settlement_date', 'england_wales_demand']])
print(f"\nTotal outliers found: {len(outliers)}")
```

```python
df.to_csv(path_out, index=False)
```
