# Creating a subset
Our dataset is really big. To make training and working with it a bit easier, we are going to take a subset of the dataset.

```python
import pandas as pd

path = "../Dataset2_Demand/4_Elec_Demand_fix_outliers.csv"
path_out = "../Dataset2_Demand/5_Elec_Demand_subset.csv"
```

### Data between 2001 & 2017
This dataset has data from 2001 until now. We will only use the data from 2001 until 2017. The other dataset ends in 2017, that is why we end there for this dataset as well.

```python
df = pd.read_csv(path)
means = df.mean(numeric_only=True) # Will be used later on

df_subset = df[(df['year'] >= 2001) & (df['year'] <= 2017)]
```

### Drop unused columns
Now that we have our subset, we are going to drop some columns that will not be useful for our model later on, or that only have NA values.

```python
total_rows = len(df_subset)

for col in df_subset.columns:
    non_na_count = df_subset[col].notna().sum()
    print(f"{col}: {non_na_count} non-NA values out of {total_rows} rows")
    
empty_cols = [c for c in df_subset.columns if df_subset[c].notna().sum() == 0]
print("\nCompletely empty columns:", empty_cols)
```

```python
columns_to_drop = ['scottish_transfer', 'ifa_flow','nsl_flow', 'eleclink_flow', 'viking_flow', 'greenlink_flow']
df_filtered = df_subset.drop(columns=columns_to_drop, errors='ignore')
```

We remove the columns nd and tsd because they dont have any impact or are not relevant for the column we want to predict.

```python
columns_to_drop = ['nd', 'tsd']
df_filtered = df_subset.drop(columns=columns_to_drop, errors='ignore')
```

We will use our mean tactic to fill any leftover NA values.

```python
means = df_filtered.mean(numeric_only=True)
df_filled = df_filtered.fillna(means)
df_filled.to_csv(path_out, index=False)
```
