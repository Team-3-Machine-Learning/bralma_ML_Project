# Restructuring dataset 1
### Renaming the columns
The column names of dataset 1 contain capital letters and spaces. We are going to rename them: removing the spaces and replacing them with underscores, and removing capital letters and replacing them with regular letters.

```python
import pandas as pd

path = "../Dataset1_UK_Housing/Raw/price_paid_records.csv"
path_out = "../Dataset1_UK_Housing/1_price_paid_records_restructured.csv"

df = pd.read_csv(path)
df.columns
```

```python
df.columns = [col.replace(' ', '_') for col in df.columns]
df.columns = [col.lower() for col in df.columns]

df.columns
```

```python
df.to_csv(path_out, index=False)
```
