# Fix outliers in dataset
### Finding price outliers
In our dataset, almost all columns are categorical. That is why we are going to fix possible outliers in the price column.

```python
import pandas as pd

path = "../Dataset1_UK_Housing/1_price_paid_records_restructured.csv" # Still use the original restructured dataset because there are no NA values
path_out = "../Dataset1_UK_Housing/3_price_paid_records_fixed_outliers.csv"
```

### Negative values
First, we use a simple check to verify if we have negative price values, which of course would not make sense.

```python
import pandas as pd

df = pd.read_csv(path)

# Tel alle negatieve prijzen
num_negative = (df["price"] < 0).sum()

print("Number of negative prices:", num_negative)
```

### Check for outliers
Now, we search for the price values that are smaller than the first quantile, or larger than the third quantile. 

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1

# Define outliers
outliers = df[(df["price"] < Q1 - 1.5 * IQR) | (df["price"] > Q3 + 1.5 * IQR)]

df.to_csv(path_out, index=False)

print("Number of outliers:", len(outliers))
```
