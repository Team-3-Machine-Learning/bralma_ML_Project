# Creating a subset
Our dataset is really big. To make training and working with it a bit easier, we are going to take a subset of the dataset.

```python
import pandas as pd

path = "../Dataset1_UK_Housing/3_price_paid_records_fixed_outliers.csv"
path_out = "../Dataset1_UK_Housing/4_price_paid_records_subset.csv"
```

### Data between 2001 & 2017
This dataset has data from 1995 until 2017. We will only use the data from 2001 until 2017. The other dataset starts in 2001, that is why we start there for this dataset as well.

```python
df = pd.read_csv(path)

df['date_of_transfer'] = pd.to_datetime(df['date_of_transfer'], errors='coerce')
df['year'] = df['date_of_transfer'].dt.year

df_subset = df[(df['year'] >= 2001) & (df['year'] <= 2017)]
```

### Drop unused columns
Now that we have our subset, we are going to drop some columns that will not be useful for our model later on. For example transaction_unique_identifier, which is an id we will not use.

```python

columns_to_drop = ['transaction_unique_identifier']
df_filtered = df_subset.drop(columns=columns_to_drop, errors='ignore')
```

After this, we check the categorical columns to see if there are any that only have one value. These will not have an impact on our model, so we drop them.

```python
cols = [
    "property_type",
    "old/new",
    "duration",
    "ppdcategory_type",
    "record_status_-_monthly_file_only"
]

for c in cols:
    print(f"--- {c} ---")
    print(df_filtered[c].unique())
    print()
```

```python
df_filtered = df_filtered.drop(columns=['record_status_-_monthly_file_only'], errors='ignore')
df_filtered.to_csv(path_out, index=False)
```
