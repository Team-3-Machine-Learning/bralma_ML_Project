# Combined notebook
This notebook contains all relevant code for the UK Housing dataset.

```python
import pandas as pd

path = "../Dataset1_UK_Housing/Raw/price_paid_records.csv"
path_out = "../Dataset1_UK_Housing/5_price_paid_records_final.csv"
df = pd.read_csv(path)

df.columns = [col.replace(' ', '_') for col in df.columns]
df.columns = [col.lower() for col in df.columns]

df['date_of_transfer'] = pd.to_datetime(df['date_of_transfer'], errors='coerce')
df['year'] = df['date_of_transfer'].dt.year

df_subset = df[(df['year'] >= 2001) & (df['year'] <= 2017)]

columns_to_drop = ['transaction_unique_identifier', 'year', 'date_of_transfer']
df_filtered = df_subset.drop(columns=columns_to_drop, errors='ignore')

df_filtered = df_filtered.drop(columns=['record_status_-_monthly_file_only'], errors='ignore')
df_filtered.to_csv(path_out, index=False)
```
